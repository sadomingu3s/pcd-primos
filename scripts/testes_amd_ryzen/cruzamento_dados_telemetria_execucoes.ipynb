```python
#Código escrito com IA (Gemini)
import pandas as pd
```

```python
path_experimentos = r"/content/results.csv"
path_dados_consumo = r"/content/mais_testes.CSV"

experimentos = pd.read_csv(path_experimentos)
dados_consumo = pd.read_csv(path_dados_consumo)
print(dados_consumo["Potência total da CPU [W]"].tail(5))
dados_consumo = dados_consumo[:-2]
print(dados_consumo["Potência total da CPU [W]"].tail(5))
dados_consumo["Potência total da CPU [W]"] = dados_consumo["Potência total da CPU [W]"].astype(float)
print(dados_consumo["Potência total da CPU [W]"].dtype)
```

```python
dados_consumo_filtrado = dados_consumo[["Date","Time","Potência total da CPU [W]"]]
print(dados_consumo_filtrado.head())
```

```python
print(experimentos.head())
```

```python
experimentos['Inicio'] = pd.to_datetime(experimentos['Inicio'])
experimentos['Fim'] = pd.to_datetime(experimentos['Fim'])

dados_consumo_filtrado["Datetime_string"] = dados_consumo_filtrado['Date'] + " " + dados_consumo_filtrado['Time']
dados_consumo_filtrado['Datetime'] = pd.to_datetime(dados_consumo_filtrado['Datetime_string'],dayfirst=True)

print(dados_consumo_filtrado.head())
```

```python
#cruzamento de dados
potencia_media = []
energias_totais = []

for index, row in experimentos.iterrows():
    inicio = row["Inicio"]
    fim = row["Fim"]
    tempo_seg = float(row["Tempo_Segundos"])

    # Filtra o df de energia pegando apenas o período de execução deste run
    mask = (dados_consumo_filtrado["Datetime"] >= inicio) & (dados_consumo_filtrado["Datetime"] <= fim)
    dados_filtrados = dados_consumo_filtrado.loc[mask]

    if not dados_filtrados.empty:
        # Tira a média de consumo em Watts
        media_w = dados_filtrados["Potência total da CPU [W]"].mean()
        # Energia (Joules) = Potência Média (W) * Tempo (s)
        energia_j = media_w * tempo_seg
    else:
        # Caso não haja capturas de energia naquele exato segundo
        media_w = None
        energia_j = None

    potencia_media.append(media_w)
    energias_totais.append(energia_j)

experimentos["Potencia_Media_Watts"] = potencia_media
experimentos["Energia_Total_Joules"] = energias_totais

print(experimentos.head())
```

```python
arquivo_saida = "resultados_experimentos_consumo.csv"
experimentos.to_csv(arquivo_saida, index=False)
```
